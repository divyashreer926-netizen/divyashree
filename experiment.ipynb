# Continual Learning RAG Demonstration

This notebook demonstrates how Retrieval-Augmented Generation (RAG) allows a small LLM to 'learn' new information without retraining.

```python
from rag_engine import RagEngine

# Initialize the engine (Make sure models are downloaded)
engine = RagEngine(persist_directory="./demo_db")
print("Engine Ready!")
```

### Step 1: Ask a question the AI doesn't know
Since the model is 'TinyLlama', it won't know specific details about your local environment or recent private events.

```python
query = "What is the secret code for the deep blue vault?"
res = engine.query(query)
print(f"AI Response: {res['answer']}")
```

### Step 2: Inject New Knowledge (Continual Learning)
We add the information to the vector database.

```python
new_info = "The secret code for the deep blue vault is 'BLUE-99X-JET'. It was set by the security team yesterday."
engine.add_knowledge(new_info)
print("Knowledge Injected.")
```

### Step 3: Ask the same question again
Now the AI should retrieve the context and answer correctly.

```python
res = engine.query(query)
print(f"AI Response (After Learning): {res['answer']}")
```
